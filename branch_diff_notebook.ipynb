# Import Required Libraries
Import necessary libraries such as GitPython for interacting with Git repositories.

```python
from git import Repo
import os
import json

repo_path = r"c:\Users\dasso\OneDrive\Desktop\Sohini-Das_assignment"
repo = Repo(repo_path)
assert not repo.bare
print(f"Repository loaded: {repo.working_tree_dir}")
```

# Connect to the Repository
Use GitPython to connect to the repository 'Sohini-Das_assignment' owned by 'Sohini-Das-Sumi'.

```python
# Fetch Branch Information
current_branch = repo.active_branch.name
default_branch = 'master'
print(f'Current branch: {current_branch}')
print(f'Default branch: {default_branch}')
```

# Calculate Diffs Between Branches
Calculate the differences (diffs) between the current branch and the default branch.

```python
base = repo.commit(default_branch)
head = repo.commit(current_branch)
diff_index = base.diff(head)
print(f'Found {len(diff_index)} files changed between {default_branch} and {current_branch}.')
```

# Compare Changes with the Default Branch
Compare and display the changes, including added, modified, and deleted files.

```python
changes = []
for diff in diff_index:
    change_type = 'modified'
    if diff.new_file:
        change_type = 'added'
    elif diff.deleted_file:
        change_type = 'deleted'
    changes.append({
        'path': diff.b_path or diff.a_path,
        'change_type': change_type,
        'a_path': diff.a_path,
        'b_path': diff.b_path,
    })
print(json.dumps(changes, indent=2))
```
